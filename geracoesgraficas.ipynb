```python
import matplotlib.pyplot as plt 
```

```python
# 1. Dados de entrada
x = [1, 2, 3, 4, 5]
y1 = [2, 5, 9, 3, 8]
y2 = [1, 2, 1, 1, 2]

# 2. Plotagem das linhas e configurações visuais 
plt.plot(x, y1, label="Relação entre X e Y1", color="red", linestyle='--', marker='o')
plt.plot(x, y2, label="Relação entre X e Y2", color="blue", linestyle='-', marker='o')

# 3. Títulos e rótulos dos eixos
plt.title("Gráfico de Linha")
plt.xlabel("VALOR X")  # Eixo horizontal
plt.ylabel("VALOR Y")  # Eixo vertical

# 4. Exibição de elementos adicionais e renderização
plt.legend()  # Exibe a legenda definida pelos 'label' em plt.plot()
plt.show()    # Renderiza o gráfico final
```

```python


# 1. Criação e plotagem dos dados
plt.scatter(x, y1, label="Relação entre X e Y1", color="red", marker='o', s=100)
plt.scatter(x, y2, label="Relação entre X e Y2", color="blue", marker='o', s=100)

# Preenchimento entre as duas séries de dados
plt.fill_between(x, y1, y2, color='orange', alpha=0.3)  # alpha adiciona transparência

# 2. Configurações visuais do gráfico
plt.title("Gráfico de Dispersão com Área Preenchida")
plt.xlabel("VALOR X (Eixo Horizontal)")
plt.ylabel("VALOR Y (Eixo Vertical)")

# Exibe a legenda configurada nos parâmetros 'label' acima
plt.legend()

# 3. Exibição do gráfico
plt.show()
```

```python
# 1. Dados de entrada
categorias = ['A', 'B', 'C', 'D']
valores = [10, 20, 15, 5]

# 2. Criando o gráfico de barras
plt.bar(categorias, valores, label='Valores das Categorias', color='skyblue')

# 3. Títulos e rótulos dos eixos
plt.title("Distribuição de Valores por Categoria")
plt.xlabel("CATEGORIAS")  # Eixo horizontal
plt.ylabel("VALORES")     # Eixo vertical

# 4. Legenda e exibição
plt.legend()
plt.show()  # Renderiza o gráfico final na tela
```

```python
# 1. Dados de entrada
categorias = ['A', 'B', 'C', 'D']
valores = [10, 20, 15, 5]

# 2. Criando o gráfico de barras horizontais
# O 'h' em plt.barh() vem de 'horizontal'. Ele inverte os eixos:
# as categorias vão para o eixo Y (vertical) e os valores para o eixo X (horizontal).
plt.barh(categorias, valores, label='Valores das Categorias', color='skyblue')

# 3. Títulos e rótulos dos eixos
plt.title("Distribuição de Valores por Categoria")
plt.xlabel("VALORES")     # Eixo X agora representa os valores (comprimento da barra)
plt.ylabel("CATEGORIAS")  # Eixo Y agora representa as categorias

# 4. Legenda e exibição
plt.legend()
plt.show()  # Renderiza o gráfico final na tela
```

```python
# 1. Dados de entrada
categorias = ['A', 'B', 'C', 'D']
valores = [10, 20, 15, 5]

# 2. Criando o gráfico de pizza (pie chart)
# plt.pie() exibe a proporção de cada valor em relação ao total.
# O parâmetro 'labels' coloca o nome de cada categoria direto nas fatias.
plt.pie(valores, labels=categorias, autopct='%1.1f%%', startangle=90)

# 3. Título do gráfico
plt.title("Proporção das Categorias")

# 4. Legenda e exibição
plt.legend(title="Categorias", loc="upper right", bbox_to_anchor=(1.2, 1))
plt.show()  # Renderiza o gráfico final na tela
```

```python
import pandas as pd
import numpy as np

# Configuração da semente para bater exatamente com os dados da aula
np.random.seed(42)

# Datas de dias úteis do primeiro semestre de 2025
datas = pd.date_range(start="2025-01-02", end="2025-06-30", freq="B")

# Simulação exata usada pelo professor
petr4 = 35.0 + np.cumsum(np.random.normal(0, 0.4, len(datas)))
itub4 = 27.0 + np.cumsum(np.random.normal(0, 0.3, len(datas)))

# Criação do DataFrame
df = pd.DataFrame({
    'Date': datas.strftime('%Y-%m-%d'),
    'PETR4': petr4,
    'ITUB4': itub4
})

# Salva o CSV com o delimitador padrão (vírgula)
df.to_csv('dados.csv', index=False)
print("Arquivo dados.csv gerado com sucesso!")
```

```python
dados = pd.read_csv('dados.csv')
```

```python
dados
```

```python
dias = dados['Date']
petro = dados['PETR4']
itau = dados['ITUB4']
fig = plt.figure(figsize=(16,10))
```

```python
plt.plot(dias, petro, label='Petro', color='green', linestyle='-', marker='.')
plt.plot(dias, itau, label='Itaú', color='Orange', linestyle='-', marker='.')
plt.title('Cotação - 2025//01')
plt.xlabel('Dias')
plt.ylabel('Valor da Ação (R$)')
plt.legend()
# Rotaciona as datas em 45 graus para não encostarem uma na outra
plt.xticks(ticks=dias[::20], labels=dias[::20], rotation=45)
plt.show()
```

```python
petro.diff() # Para entender a diferença de valores a cada dia 
```

```python
fig = plt.figure(figsize=(12, 8))

# --- Petrobras ---
# Histograma com transparência e 50 divisões (bins)
plt.hist(petro.diff().dropna(), label='Petro', color='green', alpha=0.6, bins=50) 
# Linha vertical indicando a média da variação
plt.axvline(petro.diff().dropna().mean(), color='green', linestyle='--', linewidth=4) 

# --- Itaú ---
plt.hist(itau.diff().dropna(), label='Itaú', color='orange', alpha=0.6, bins=50) 
plt.axvline(itau.diff().dropna().mean(), color='orange', linestyle='--', linewidth=4) 

# --- Formatação ---
plt.title('Variação Diária - 2025/01')
plt.xlabel('Variação (R$)')
plt.ylabel('Frequência (Dias)')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Estilo Dark Background
plt.style.use('dark_background')

fig, (ax_cima, ax_baixo) = plt.subplots(figsize=(12, 8), nrows=2, sharex=True)

# --- Gráfico de Cima (Petrobras) ---
ax_cima.hist(petro.diff().dropna(), label='Petrobras', color='green', bins=50)
ax_cima.axvline(petro.diff().dropna().mean(), color='gray', linestyle='--', linewidth=2.5)
ax_cima.grid(True, linestyle=':', alpha=0.6)
ax_cima.set_ylabel('Frequência (Dias)')
ax_cima.legend(loc='upper right')

# --- Gráfico de Baixo (Itaú) ---
ax_baixo.hist(itau.diff().dropna(), label='Itaú', color='orange', bins=50)
ax_baixo.axvline(itau.diff().dropna().mean(), color='gray', linestyle='--', linewidth=2.5)
ax_baixo.grid(True, linestyle=':', alpha=0.6)
ax_baixo.set_xlabel('Variação (R$)')
ax_baixo.set_ylabel('Frequência (Dias)')
ax_baixo.legend(loc='upper right')

# --- Título e Ajustes Finais ---
fig.suptitle('Variação Diária das Ações - 2025/1', fontsize=14)

# Organiza o espaçamento entre os subplots automaticamente
plt.tight_layout()
plt.show()
```

```python
# Aplicando o estilo dark
plt.style.use('dark_background')

fig = plt.figure(figsize=(12, 8))
grid = fig.add_gridspec(2, 2)

# --- Subplots ---
ax_linha = fig.add_subplot(grid[0, :])   # Ocupa toda a primeira linha (linha 0, colunas 0 e 1)
ax_hist_1 = fig.add_subplot(grid[1, 0])  # Segunda linha, primeira coluna
ax_hist_2 = fig.add_subplot(grid[1, 1])  # Segunda linha, segunda coluna

# --- 1. Gráfico de Linhas (Cotação) ---
ax_linha.plot(dias, petro, label='Petro', color='green', linestyle='-', marker='.')
ax_linha.plot(dias, itau, label='Itaú', color='orange', linestyle='-', marker='.')
ax_linha.set_title('Cotação Histórica - 2025/01')
ax_linha.set_xlabel('Dias')
ax_linha.set_ylabel('Valor da Ação (R$)')
ax_linha.set_xticks(ticks=dias[::20])
ax_linha.legend(loc='upper right')
ax_linha.grid(True, linestyle=':', alpha=0.5)

# --- 2. Histograma Petrobras ---
ax_hist_1.hist(petro.diff().dropna(), color='green', bins=50) 
ax_hist_1.axvline(petro.diff().dropna().mean(), color='gray', linestyle='--', linewidth=2.5)
ax_hist_1.grid(True, linestyle=':', alpha=0.5)
ax_hist_1.set_title('Variação Diária - Petrobras')
ax_hist_1.set_xlabel('Variação (R$)')
ax_hist_1.set_ylabel('Frequência (Dias)')

# --- 3. Histograma Itaú ---
ax_hist_2.hist(itau.diff().dropna(), color='orange', bins=50) 
ax_hist_2.axvline(itau.diff().dropna().mean(), color='gray', linestyle='--', linewidth=2.5)
ax_hist_2.grid(True, linestyle=':', alpha=0.5)
ax_hist_2.set_title('Variação Diária - Itaú')
ax_hist_2.set_xlabel('Variação (R$)')
ax_hist_2.set_ylabel('Frequência (Dias)')

# --- Título Geral e Ajuste de Espaçamento ---
fig.suptitle('Análise Comparativa de Ações (PETR4 vs ITUB4)', fontsize=14)

# rect=[left, bottom, right, top] evita que o tight_layout encoste no suptitle
fig.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```
